```python
from groq import Groq
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
GROQ_API_KEY = os.getenv("GROQ_API_KEY")
if not GROQ_API_KEY:
    raise RuntimeError("Missing GROQ_API_KEY. Add it to the project .env file (see .env.example).")
client = Groq(api_key= GROQ_API_KEY)
```

```python
def ask_llm(system_prompt, user_prompt):
    message = [
        {
            "role":"system",
            "content": system_prompt
        },
        {
            "role":"user",
            "content":user_prompt
        }
    ]
    chat_completion = client.chat.completions.create(messages = message, model="openai/gpt-oss-120b")
    response = chat_completion.choices[0].message.content
    return response
```

```python
patient_name = input("Enter patient name: ")
patient_notes = input("Enter your notes: ")
```

```python
system_prompt="You are a doctor and you will get patient's raw consultation notes and generate a clean doctor's summary. The Summary with fixed sections are Symptoms, Diagnosis, Recommendation - in the same format every time, regardless of input."
user_prompt = "My Name is {name} and my notes are {notes}"
```

```python
print(ask_llm(system_prompt,user_prompt.format(name = patient_name,notes = patient_notes)))
```
